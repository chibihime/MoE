# DEAP Dataset Exploratory Data Analysis

This notebook loads the DEAP dataset and performs exploratory analysis, including summary statistics and visualizations for EEG, physiological, and label data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import sys
sys.path.append(os.path.join(os.getcwd(), 'src', 'data'))
from acquisition import load_deap

# Set plot style
sns.set(style='whitegrid')
```

```python
# Path to DEAP dataset
DATA_DIR = './data/deap-dataset/data_preprocessed_python'

data = load_deap(DATA_DIR)

eeg_dfs = data['eeg']
physio_dfs = data['physio']
labels_df = data['labels']
```

```python
# Inspect loaded dataset
print('Keys:', data.keys())

# Preview EEG data
if eeg_dfs:
    print('EEG DataFrame shape:', eeg_dfs[0].shape)
    display(eeg_dfs[0].head())
else:
    print('No EEG data loaded.')

# Preview Physio data
if physio_dfs:
    print('Physio DataFrame shape:', physio_dfs[0].shape)
    display(physio_dfs[0].head())
else:
    print('No Physio data loaded.')

# Preview Labels
if labels_df is not None:
    print('Labels DataFrame shape:', labels_df.shape)
    display(labels_df.head())
else:
    print('No labels data loaded.')
```

```python
# Summary statistics
if eeg_dfs:
    print('EEG Summary Statistics:')
    display(eeg_dfs[0].describe())
if physio_dfs:
    print('Physio Summary Statistics:')
    display(physio_dfs[0].describe())
if labels_df is not None:
    print('Labels Summary Statistics:')
    display(labels_df.describe())
```

```python
# Visualize EEG Data
if eeg_dfs:
    eeg_sample = eeg_dfs[0]
    plt.figure(figsize=(12, 4))
    plt.plot(eeg_sample.iloc[:, 0], label='EEG Channel 1')
    plt.title('Sample EEG Channel (Subject 1)')
    plt.xlabel('Time')
    plt.ylabel('Amplitude')
    plt.legend()
    plt.show()
else:
    print('No EEG data to visualize.')
```

```python
# Visualize Physiological Data
if physio_dfs:
    physio_sample = physio_dfs[0]
    plt.figure(figsize=(12, 4))
    plt.plot(physio_sample.iloc[:, 0], label='Physio Channel 1')
    plt.title('Sample Physiological Channel (Subject 1)')
    plt.xlabel('Time')
    plt.ylabel('Value')
    plt.legend()
    plt.show()
else:
    print('No physiological data to visualize.')
```

```python
# Analyze Labels Distribution
if labels_df is not None:
    plt.figure(figsize=(10, 4))
    plt.subplot(1, 2, 1)
    sns.histplot(labels_df['valence'], bins=20, kde=True)
    plt.title('Valence Distribution')
    plt.xlabel('Valence')
    plt.subplot(1, 2, 2)
    sns.histplot(labels_df['arousal'], bins=20, kde=True)
    plt.title('Arousal Distribution')
    plt.xlabel('Arousal')
    plt.tight_layout()
    plt.show()
else:
    print('No labels data to analyze.')
```
